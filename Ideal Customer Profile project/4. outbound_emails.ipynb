# 4 — Three emails

A ranked list is an artefact. **An email is the job.**

Each of these is written to one of the top three accounts and cites a specific fact drawn from the
data about *that company* — not a merge field, and not a fact that would be true of any developer.
The test each has to pass: **could this email have been sent to anyone else on the list?** If yes, it
is a mail merge and it belongs in the bin.

All three follow the same shape, which is the shape of a discovery call compressed into six lines:
observation the recipient can verify, the inference that observation supports, the cost of that
inference being true, and one low-commitment ask.

## Why these are not mail merges

Apply the test. Email 1 turns on *group accounts unusually early*. Email 2 turns on *32 entities*.
Email 3 turns on *PBSA catchment economics*. Swap any one onto a different recipient and the opening
line becomes false — which is the whole point.

Every fact used is free, public, and machine-readable at segment scale. Nothing here required a data
provider, a scraper or a subscription; the entire personalisation budget was one monthly CSV.
